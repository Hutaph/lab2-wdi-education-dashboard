## IV. Phân tích dữ liệu thiếu

Trong dữ liệu WDI, giá trị thiếu được biểu diễn bằng ký hiệu `".."`. Để tính toán tỷ lệ thiếu dữ liệu, ta tạo một bản sao tạm thời và thay `".."` bằng `NaN`.

```python
df_missing = df_valid.copy()
df_missing[year_cols] = df_missing[year_cols].replace("..", np.nan)
```

```python
def plot_missing_values(
    df,
    label_col,
    percentage_col="Tỷ lệ thiếu (%)",
    title="Tỷ lệ khuyết thiếu dữ liệu",
    xlabel="Tỷ lệ khuyết thiếu (%)",
    ylabel="",
    figsize=(14, 8),
    horizontal=True
):
    import matplotlib.pyplot as plt
    import seaborn as sns

    plot_df = df.copy()

    # Đảm bảo percentage là numeric
    if plot_df[percentage_col].dtype == object:
        plot_df[percentage_col] = (
            plot_df[percentage_col]
            .astype(str)
            .str.replace("%", "", regex=False)
        )

    plot_df[percentage_col] = pd.to_numeric(
        plot_df[percentage_col],
        errors="coerce"
    )

    # Sort giảm dần để dễ nhìn
    plot_df = plot_df.sort_values(by=percentage_col, ascending=False)

    sns.set_theme(style="whitegrid")
    plt.figure(figsize=figsize)

    if horizontal:
        ax = sns.barplot(
            data=plot_df,
            x=percentage_col,
            y=label_col,
            color="#d95f5f"
        )

        plt.xlabel(xlabel, fontsize=12, fontweight="bold")
        plt.ylabel(ylabel if ylabel else label_col, fontsize=12, fontweight="bold")

        for p in ax.patches:
            width = p.get_width()
            ax.text(
                width + 1,
                p.get_y() + p.get_height() / 2,
                f"{width:.2f}%",
                va="center",
                fontsize=9,
                fontweight="bold"
            )

        plt.xlim(0, 100)

    else:
        ax = sns.barplot(
            data=plot_df,
            x=label_col,
            y=percentage_col,
            color="#d95f5f"
        )

        plt.xlabel(ylabel if ylabel else label_col, fontsize=12, fontweight="bold")
        plt.ylabel(xlabel, fontsize=12, fontweight="bold")
        plt.xticks(rotation=45)

        for p in ax.patches:
            height = p.get_height()
            ax.text(
                p.get_x() + p.get_width() / 2,
                height + 1,
                f"{height:.2f}%",
                ha="center",
                va="bottom",
                fontsize=9,
                fontweight="bold"
            )

        plt.ylim(0, 100)

    plt.title(title, fontsize=16, fontweight="bold", pad=20)
    plt.tight_layout()
    plt.show()
```

### 4. Missing theo chỉ số

```python
missing_by_indicator = []

for _, row in df_missing.iterrows():
    missing_count = row[year_cols].isna().sum()
    available_count = row[year_cols].notna().sum()
    total_count = missing_count + available_count
    
    missing_by_indicator.append({
        "Tên chỉ số": row["Series Name"],
        "Mã chỉ số": row["Series Code"],
        "Quốc gia": row["Country Name"],
        "Số ô thiếu": missing_count,
        "Số ô có dữ liệu": available_count,
        "Tổng số ô": total_count
    })

missing_by_indicator = pd.DataFrame(missing_by_indicator)

missing_by_indicator_summary = (
    missing_by_indicator
    .groupby(["Tên chỉ số", "Mã chỉ số"], as_index=False)
    .agg({
        "Số ô thiếu": "sum",
        "Số ô có dữ liệu": "sum",
        "Tổng số ô": "sum"
    })
)

missing_by_indicator_summary["Tỷ lệ thiếu (%)"] = (
    missing_by_indicator_summary["Số ô thiếu"] /
    missing_by_indicator_summary["Tổng số ô"] * 100
).round(2)

missing_by_indicator_summary = missing_by_indicator_summary.sort_values(
    "Tỷ lệ thiếu (%)",
    ascending=False
)

missing_by_indicator_summary
```

```python
indicator_short_names = {
    "Government expenditure per student, secondary (% of GDP per capita)": "Chi tiêu/học sinh trung học",
    "Literacy rate, adult total (% of people ages 15 and above)": "Tỷ lệ biết chữ người lớn",
    "Children out of school, primary, female": "Trẻ ngoài trường, nữ",
    "Children out of school, primary, male": "Trẻ ngoài trường, nam",
    "School enrollment, primary and secondary (gross), gender parity index (GPI)": "Chỉ số cân bằng giới",
    "Government expenditure on education, total (% of GDP)": "Chi giáo dục (% GDP)",
    "Government expenditure on education, total (% of government expenditure)": "Chi giáo dục (% ngân sách)",
    "Primary completion rate, total (% of relevant age group)": "Hoàn thành tiểu học",
    "Lower secondary completion rate, total (% of relevant age group)": "Hoàn thành THCS",
    "Children out of school, primary": "Trẻ ngoài trường",
    "School enrollment, secondary (% gross)": "Nhập học trung học",
    "School enrollment, tertiary (% gross)": "Nhập học đại học",
    "School enrollment, primary (% gross)": "Nhập học tiểu học",
    "GDP per capita (current US$)": "GDP bình quân đầu người",
    "Life expectancy at birth, total (years)": "Tuổi thọ",
    "Urban population (% of total population)": "Dân số đô thị",
    "Population, total": "Dân số"
}

missing_by_indicator_summary["Tên chỉ số ngắn"] = (
    missing_by_indicator_summary["Tên chỉ số"]
    .map(indicator_short_names)
    .fillna(missing_by_indicator_summary["Tên chỉ số"])
)

plot_missing_values(
    df=missing_by_indicator_summary,
    label_col="Tên chỉ số ngắn",
    percentage_col="Tỷ lệ thiếu (%)",
    title="Tỷ lệ khuyết thiếu dữ liệu theo chỉ số",
    xlabel="Tỷ lệ khuyết thiếu (%)",
    ylabel="Chỉ số",
    figsize=(12, 8),
    horizontal=True
)
```

**Liên hệ với preprocessing:**  
Từ kết quả EDA, pipeline chọn giai đoạn 2001–2024 và loại năm 2025 do thiếu dữ liệu quá cao. Missing values được xử lý theo nhóm chỉ số: nội suy mượt cho chỉ số bối cảnh, nội suy khoảng trống ngắn cho enrollment/completion/finance, log-transform cho nhóm trẻ ngoài trường học, suy ra tổng/nam/nữ khi có đủ thành phần, rồi fallback bằng median cùng chỉ số khi cần.

### 1. Thống kê theo từng chỉ số

```python
summary_rows = []

for _, row in df_stats.iterrows():
    values = row[year_cols].dropna()
    
    summary_rows.append({
        "Tên chỉ số": row["Series Name"],
        "Mã chỉ số": row["Series Code"],
        "Quốc gia": row["Country Name"],
        "Số giá trị": values.count(),
        "Trung bình": values.mean(),
        "Trung vị": values.median(),
        "Nhỏ nhất": values.min(),
        "Lớn nhất": values.max(),
        "Độ lệch chuẩn": values.std()
    })

summary_df = pd.DataFrame(summary_rows)

summary_by_indicator = (
    summary_df
    .groupby(["Tên chỉ số", "Mã chỉ số"], as_index=False)
    .agg({
        "Số giá trị": "sum",
        "Trung bình": "mean",
        "Trung vị": "median",
        "Nhỏ nhất": "min",
        "Lớn nhất": "max",
        "Độ lệch chuẩn": "mean"
    })
)

summary_by_indicator
```

**Nhận xét:**  
Các chỉ số có thang đo rất khác nhau: phần trăm, số người, USD, số năm và chỉ số tỷ lệ. Nhóm bối cảnh phát triển chênh lệch mạnh về dân số và GDP bình quân đầu người; nhóm giáo dục có một số giá trị vượt 100% do đặc thù gross enrollment/completion; nhóm trẻ ngoài trường là số tuyệt đối nên cần so sánh kèm quy mô dân số.


### 2. Kiểm tra giá trị lớn hơn 100%

```python
percent_indicators = [
    "School enrollment, primary (% gross)",
    "School enrollment, secondary (% gross)",
    "School enrollment, tertiary (% gross)",
    "Primary completion rate, total (% of relevant age group)",
    "Lower secondary completion rate, total (% of relevant age group)",
    "Literacy rate, adult total (% of people ages 15 and above)",
    "Government expenditure on education, total (% of GDP)",
    "Government expenditure on education, total (% of government expenditure)",
    "Government expenditure per student, secondary (% of GDP per capita)",
    "Urban population (% of total population)"
]
```

```python
percent_summary = summary_by_indicator[
    summary_by_indicator["Tên chỉ số"].isin(percent_indicators)
].copy()

percent_summary[
    percent_summary["Lớn nhất"] > 100
].sort_values("Lớn nhất", ascending=False)
```

**Nhận xét:**  
Có 5 chỉ số phần trăm có giá trị lớn nhất vượt 100%: nhập học tiểu học, trung học, đại học, hoàn thành tiểu học và hoàn thành THCS. Đây không nhất thiết là lỗi dữ liệu vì các chỉ số gross có thể bao gồm người học ngoài độ tuổi chính thức hoặc học lại; khi trực quan hóa cần ghi chú cách diễn giải này.
